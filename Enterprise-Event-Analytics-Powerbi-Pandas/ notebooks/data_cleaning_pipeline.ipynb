```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("messy_event_ledger_2026.csv")
```

```python
df.info()
```

```python
df.dtypes
```

```python
df[~df['OrderID'].astype(str).str.isdigit()]
```

```python
df['id_length'] = df['OrderID'].astype(str).str.len()
df['OrderID'] = df['OrderID'].str.strip()
print(df['id_length'].value_counts())
```

```python
df['AttendeeProfile'].value_counts(dropna=False)
```

```python
df['AttendeeProfile'] = df['AttendeeProfile'].str.strip().str.title()
df['AttendeeProfile'] = df['AttendeeProfile'].replace({
    'Social activity': 'Social Activity',
    'Social_Activity': 'Social Activity'
})
```

```python
df['AttendeeProfile'].unique()
```

```python
df['EventType'].value_counts(dropna=False)
```

```python
df['EventType'] = df['EventType'].str.strip().str.title()
df['EventType'].value_counts(dropna=False)
```

```python
df['EventName'].value_counts(dropna=False)
```

```python
df['EventCity'].value_counts(dropna=False)
```

```python
df['BasePrice_INR'].value_counts()
```

```python
df['BasePrice_INR'] = df['BasePrice_INR'].astype(str).str.replace(' ', '', regex=False)
df['BasePrice_INR'] = df['BasePrice_INR'].str.replace('INR', '', regex=False)
print(df['BasePrice_INR'].value_counts())
```

```python
df['DiscountAmount_INR'].value_counts()
```

```python
df['TicketSellingPrice_INR'] = pd.to_numeric(df['TicketSellingPrice_INR'], errors='coerce')
df['TicketSellingPrice_INR'].value_counts()
```

```python
df['NetRevenue_INR'] = pd.to_numeric(df['NetRevenue_INR'], errors='coerce')
df["NetRevenue_INR"].value_counts()
```

```python
df['CheckInStatus'].value_counts()
```

```python
mapping_rules = {
    'Yes': 'Attended', 'Y': 'Attended', '1': 'Attended', 'true': 'Attended', 'True': 'Attended',
    'No': 'Not Attended', 'N': 'Not Attended', '0': 'Not Attended', 'false': 'Not Attended', 'False': 'Not Attended'
}

df['CheckInStatus'] = df['CheckInStatus'].astype(str).str.strip().replace(mapping_rules)
df['CheckInStatus'].value_counts()
```

```python
df['OrderTimestamp'].value_counts()
```

```python
df['OrderTimestamp'] = df['OrderTimestamp'].replace('00-00-2026', np.nan)
df['OrderTimestamp'] = pd.to_datetime(df['OrderTimestamp'], format='mixed', errors='coerce')
df['OrderTimestamp'] = df['OrderTimestamp'].dt.date
default_date = pd.to_datetime('2026-01-01').date()
df['OrderTimestamp'] = df['OrderTimestamp'].fillna(default_date)
print(df['OrderTimestamp'].value_counts())
```

```python
df['DiscountAmount_INR'] = df['DiscountAmount_INR'].fillna(0)
df['TicketSellingPrice_INR'] = df['TicketSellingPrice_INR'].fillna(0)
df['NetRevenue_INR'] = df['NetRevenue_INR'].fillna(0)
```

```python
df.to_csv("clean_event_ledger.csv", index=False)
```
